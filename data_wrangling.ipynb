### to download any parquet file from internet 

```python
  
import requests
import pandas as pd
import time 
import logging

date = '2024-02'
url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{date}.parquet"
print(f'Downloading Green taxi file for :  {date}')

try : 
    response = requests.get(url, stream=True)
    if response.status_code == 200:
        with open(f"yellow_tripdata_{date}.parquet", 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"Download complete: green_tripdata_{date}.parquet")
        time.sleep(2)
    else:
        print(f'Download failed: {response.status_code}')
        
except Exception as e : 
    print(f"ingestion has failed. Error : {e}")
```
